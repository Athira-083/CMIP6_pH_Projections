```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(14, 6), sharex=True, sharey=True)

regions = ['(a) Arabian Sea', '(b) Bay of Bengal', '(c) Equatorial Indian Ocean', '(d) South Indian Ocean']
datasets = [
    (avg_subset_ds1, avg_subset_ds2, avg_subset_ds3, avg_subset_ds4),
    (avg_subset_ds1_bob, avg_subset_ds2_bob, avg_subset_ds3_bob, avg_subset_ds4_bob),
    (avg_subset_ds1_cio, avg_subset_ds2_cio, avg_subset_ds3_cio, avg_subset_ds4_cio),
    (avg_subset_ds1_sio, avg_subset_ds2_sio, avg_subset_ds3_sio, avg_subset_ds4_sio)
]

uncertainties = [
    (avg_model_Ut_hist_as, avg_model_Ut_126_as, avg_model_Ut_245_as, avg_model_Ut_585_as),
    (avg_model_Ut_hist_bob, avg_model_Ut_126_bob, avg_model_Ut_245_bob, avg_model_Ut_585_bob),
    (avg_model_Ut_hist_cio, avg_model_Ut_126_cio, avg_model_Ut_245_cio, avg_model_Ut_585_cio),
    (avg_model_Ut_hist_sio, avg_model_Ut_126_sio, avg_model_Ut_245_sio, avg_model_Ut_585_sio)
]

labels = ['Historical', 'SSP1-2.6', 'SSP2-4.5', 'SSP5-8.5']
colors = ['black', 'lightgreen', 'blue', 'red']

subplot_labels = ['(a)', '(b)', '(c)', '(d)']
import xarray as xr

for ax, label, region, data, uncertainty in zip(axes.flatten(), subplot_labels, regions, datasets, uncertainties):
    for dataset, uncert, color, dataset_label in zip(data, uncertainty, colors, labels):
        ph_data = dataset['ph']
        year = dataset['year'].values  
        
        uncert_data = uncert['ph'] 

#         if uncert_data.shape != ph_data.shape:
#             uncert_data = uncert_data.interp(year=ph_data['year'], method="linear") 
#             print(f"Interpolated Uncertainty data shape: {uncert_data.shape}")
        
        ax.plot(year, ph_data, label=dataset_label, color=color, linewidth=2)

        uncert_values = uncert_data.values 
        
        ax.fill_between(
            year,
            ph_data - uncert_values* 4.1,  
            ph_data + uncert_values* 4.1,  
            color=color,
            alpha=0.16
        )

    ax.text(0.02, 0.05, label, transform=ax.transAxes, fontsize=14, fontweight='bold')
    ax.grid(False)
    ax.tick_params(axis='both', which='major', labelsize=12, width=1, length=5, pad=10)
    for tick in ax.get_xticklabels():
        tick.set_fontweight("bold")
    for tick in ax.get_yticklabels():
        tick.set_fontweight("bold")
    ax.set_xlim(1985, 2100)
    plt.xticks(np.arange(1990, 2101, 20))  # decade ticks only

    ax.set_ylim(7.6, 8.2)
    ax.yaxis.set_ticks(np.arange(7.6, 8.3, 0.2))
fig.text(0.065, 0.5, 'pH', va='center', rotation='vertical', fontsize=17, fontweight='bold')
plt.subplots_adjust(top=0.9, wspace=0.1, hspace=0.15)
# plt.suptitle('Bias-corrected multi-model Ensemble pH (1985-2100)', fontsize=16, fontweight='bold')
plt.legend(loc='upper center', bbox_to_anchor=(-0.05, -0.2), ncol=4, fontsize=12)  
plt.savefig(
    "timeseries_regional_pH_ens_v1.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
