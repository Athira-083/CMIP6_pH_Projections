```python
import numpy as np
import xarray as xr
import matplotlib as plt
```

```python
ds1=xr.open_dataset('H+_Modeluncertainty_hist.nc')

ds2=xr.open_dataset('UIN_STL_hist_H+.nc')


ds1= ds1.rename({"ph": "H+"})

ds2 = ds2.rename({"internal_variability": "H+"})
```

```python
total_uncertainty= np.sqrt(ds1**2 + ds2**2)
```

```python
total_uncertainty['H+'].plot()
```
