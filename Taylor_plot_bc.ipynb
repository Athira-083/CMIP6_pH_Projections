```python
import skill_metrics as sm
import matplotlib.pyplot as plt
from matplotlib import rcParams
import numpy as np
plt.rcParams['axes.labelweight'] = 'bold'
def Taylor_diag(a,ax,num):
    MARKERS = { "CMIP6_ENS_pH": {
            "labelColor": "k",
           "symbol": "*",
            "size": 16,
            "faceColor": "b",
            "edgeColor": "b",
        },

           "BEM_pH": {
            "labelColor": "k",
            "symbol": "*",
            "size": 16,
            "faceColor": "r",
            "edgeColor": "r",
        }
            }

    # plt.rcParams['font.weight']='bold'
    plt.rcParams['font.weight'] = 'bold'
    plt.rcParams['axes.labelweight'] = 'bold'

    rcParams["figure.figsize"] = [12.0, 13.4]
    rcParams['lines.linewidth'] = 1.5 # line width for plots
    rcParams['lines.dashed_pattern'] = [3.7, 4.5]

    rcParams.update({'font.size': 12}) # font size of axes text

    sdev = np.array(a[0])
    crmsd= np.array(a[1])
    ccoef = np.array(a[2])

    rm=[0.003, 0.006, 0.009]
    rm1=[0.002, 0.004, 0.006, 0.008]
    rm2=[0.002, 0.004, 0.006, 0.008]
    rm3=[0.004, 0.008, 0.012, 0.016]
    rmm=[rm,rm1,rm2,rm3]
#     st=[0.0, 0.005, 0.01, 0.015, 0.02]
#     st1=[0.0, 0.005, 0.01,0.015]
#     st2=[0.0, 0.005, 0.01,0.015]
#     st3=[0.0,0.01,0.02,0.03]
#     std=[st,st1,st2,st3]
    
    if num == 0:
        sm.taylor_diagram(ax, sdev, crmsd, ccoef, markerobs='*', colobs='m', titleOBS='TAKAHASHI', widthRMS=1.5, 
                          tickRMSangle=110, showlabelsRMS='on', markers=MARKERS, markerLegend='on', 
                          titleRMSDangle=80.0, markerSize=20, colRMS= 'black', titleRMS='off', tickRMS=rmm[0],colCOR= 'black' , titleCOR='off', tickCOR=[0,0.2,0.4,0.6,0.8,0.9,0.95,0.99])
    elif num == 1:
        sm.taylor_diagram(ax, sdev, crmsd, ccoef, markerobs='*', colobs='m', titleOBS='TAKAHASHI', widthRMS=1.5, 
                          tickRMSangle=115, showlabelsRMS='on', markers=MARKERS, markerLegend='on', 
                          titleRMSDangle=120.0, markerSize=20, colRMS= 'black', titleRMS='off', tickRMS=rmm[1],colCOR= 'black' , titleCOR='off', tickCOR=[0,0.2,0.4,0.6,0.8,0.9,0.95,0.99])
    elif num == 2:
        sm.taylor_diagram(ax, sdev, crmsd, ccoef, markerobs='*', colobs='m', titleOBS='TAKAHASHI', widthRMS=1.5, 
                          tickRMSangle=120, showlabelsRMS='on', markers=MARKERS, markerLegend='on', 
                          titleRMSDangle=80.0, markerSize=20, colRMS= 'black', titleRMS='off', tickRMS=rmm[2],colCOR= 'black' , titleCOR='off', tickCOR=[0,0.2,0.4,0.6,0.8,0.9,0.95,0.99])
    elif num==3:
        sm.taylor_diagram(ax, sdev, crmsd, ccoef, markerobs='*', colobs='m', titleOBS='TAKAHASHI', widthRMS=1.5, 
                          tickRMSangle=135, showlabelsRMS='on', markers=MARKERS, markerLegend='on', 
                          titleRMSDangle=145.0, markerSize=20, colRMS= 'black', titleRMS='off', tickRMS=rmm[3],colCOR= 'black' , titleCOR='off', tickCOR=[0,0.2,0.4,0.6,0.8,0.9,0.95,0.99])
    else:
        print('nothing.........')
    
    
###############################################################################

from sklearn.metrics import mean_squared_error as ms
sdev_as = [0.006, 0.009, 0.007]
crmsd_as = [0.0, 0.013, 0.015]
ccoef_as = [1.0, 0.38, 0.80]
taylor_AS=[sdev_as,crmsd_as,ccoef_as]


sdev_bob = [0.005, 0.009, 0.006]
crmsd_bob = [0.0, 0.016, 0.019]
ccoef_bob = [1.0, 0.65, 0.78]
taylor_BOB=[sdev_bob,crmsd_bob,ccoef_bob]


sdev_cio = [0.007, 0.009, 0.006]
crmsd_cio = [0.0, 0.011, 0.015]
ccoef_cio = [1.0, 0.83,0.92]
taylor_CIO=[sdev_cio,crmsd_cio,ccoef_cio]


sdev_sio = [0.014, 0.0205, 0.016]
crmsd_sio = [0.0, 0.015, 0.018]
ccoef_sio = [1.0, 0.91, 0.95]
taylor_SIO=[sdev_sio,crmsd_sio,ccoef_sio]

###############################################################################

fig,ax=plt.subplots(2,2,figsize=(18,15))
ax=ax.ravel()
# fig.delaxes(ax[3])import numpy as np

# fig.delaxes(ax[2])
# fig.delaxes(ax[1])


Taylor_diag(taylor_AS,ax[0],0)
Taylor_diag(taylor_BOB,ax[1],1)
Taylor_diag(taylor_CIO,ax[2],2)
Taylor_diag(taylor_SIO,ax[3],3)
plt.subplots_adjust(wspace=0.05,hspace=0.30)

ax[0].set_title('(a)    AS',fontsize=18,pad=40,fontweight='bold', c='k')
ax[1].set_title('(b)    BOB',fontsize=18,pad=40,fontweight='bold', c='k')
ax[2].set_title('(c)    EIO',fontsize=18,pad=40,fontweight='bold', c='k')
ax[3].set_title('(d)    SIO',fontsize=18,pad=40,fontweight='bold', c='k')

for i in range(4):
#     ax[i].text(0.05, -0.1, f'({chr(97 + i)})', transform=ax[i].transAxes, fontsize=14, va='top', ha='right')
    ax[i].tick_params(axis='x', colors='black', labelsize=14, )  # Change x-axis color and font
    ax[i].yaxis.get_label().set_color('black')  # Change y-axis label color
    ax[i].tick_params(axis='y', colors='black', labelsize=14)  # Change y-axis color and font
#for i in range(4):
    #ax[i].text(0.05, -0.1, f'({chr(97 + i)})', transform=ax[i].transAxes,
                #fontsize=14, va='top', ha='right')
import matplotlib as mpl
from matplotlib import rcParams
mpl.rc('legend', borderpad=0.4, loc='upper center') # default is 0.4
# Change vertical space between the legend entries
mpl.rc('legend', labelspacing=0.5) #default is 0.5
mpl.rc('legend', fontsize='large')
mpl.rc('legend', loc='best')
# plt.rc('legend', fontsize='large', title_fontsize='large', shadow=False, fancybox=False)
mpl.rc('legend', )
mpl.rc('legend', columnspacing=2.0) #default is 2.0
mpl.rc('legend', shadow=False) #default is False
plt.savefig('taylor_diagram_review_f2.png', bbox_inches = 'tight', dpi = 600)

# plt.show()

```
