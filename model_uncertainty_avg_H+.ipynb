```python
import numpy as np
import xarray as xr
import matplotlib as plt
```

```python
ds1=xr.open_dataset('H+_std of ens models_ssp585_r1.nc')
ds1_period = ds1.sel(time=slice('1985', '2014'))
```

```python
kk = 0
for i in range(len(ds1_period.time)):
    kk=kk+(ds1_period.isel(time=i)) ** 2

kk1 = np.sqrt(kk/ len(ds1_period.time))
```

```python
kk1.ph.plot()
```

```python
kk1.to_netcdf('H+_Modeluncertainty_hist.nc')
```
