```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.cm import get_cmap
from matplotlib.colors import Normalize

plt.rcParams['font.weight'] = 'bold'

subplot_labels = ['(a)', '(b)', '(c)', '(d)']

# Model names
models = [
    'CanESM5', 'CESM2', 'CESM2-WACCM', 'CMCC-ESM2',
    'GFDL-ESM4', 'IPSL-CM6A-LR', 'MPI-ESM1-2-HR',
    'NorESM2-MM', 'NorESM2-LM', 'CMIP6_ENS_pH'
]

# Taylor Skill Scores
TSS = {
    'AS':  [0.31, 0.40, 0.40, 0.43, 0.32, 0.30, 0.34, 0.30, 0.30, 0.43],
    'BOB': [0.35, 0.55, 0.53, 0.50, 0.34, 0.53, 0.33, 0.45, 0.38, 0.61],
    'EIO': [0.59, 0.56, 0.51, 0.65, 0.61, 0.58, 0.58, 0.54, 0.69, 0.77],
    'SIO': [0.71, 0.80, 0.81, 0.80, 0.68, 0.66, 0.74, 0.79, 0.70, 0.76]
}
# Colormap
cmap = get_cmap('YlGn')

# Create subplots
fig, axes = plt.subplots(2, 2, figsize=(20, 12))
axes = axes.flatten()

# Plot each region
for idx_ax, (ax, (region, scores)) in enumerate(zip(axes, TSS.items())):
    scores = np.array(scores)

    # Sort models by skill score (low → high)
    idx = np.argsort(scores)
    sorted_scores = scores[idx]
    sorted_models = np.array(models)[idx]

    # Normalize colors within region
    norm = Normalize(vmin=sorted_scores.min(),
                     vmax=sorted_scores.max())
    bar_colors = cmap(norm(sorted_scores))

    x = np.arange(len(sorted_scores))
    ax.bar(x, sorted_scores, color=bar_colors, width=0.8)

    # Value labels
    for i, val in enumerate(sorted_scores):
        ax.text(i, val + 0.01, f'{val:.2f}',
                ha='center', va='bottom',
                fontsize=14.5, fontweight='bold')

    ax.set_title(region, fontsize=22, fontweight='bold')
    ax.set_ylim(0, 1)

    ax.set_xticks(x)
    ax.set_xticklabels(sorted_models, rotation=36, ha='right',
                       fontsize=14, fontweight='bold')

    # Y-label only for left column
    # if idx_ax % 2 == 0:
    #     ax.set_ylabel('Taylors Skill Score', fontsize=22, fontweight='bold')
    # else:
    #     ax.set_ylabel('')
    #     ax.tick_params(axis='y', labelleft=False)
    fig.text(-0.01, 0.5, 'Taylor Skill Score',
         va='center', rotation='vertical',
         fontsize=24, fontweight='bold')
    ax.tick_params(axis='y', labelsize=18)

    # Subplot labels (a, b, c, d)
    ax.text(0.02, 0.95, subplot_labels[idx_ax],
            transform=ax.transAxes,
            fontsize=20, fontweight='bold',
            va='top', ha='left')

    # Bold subplot borders
    for spine in ax.spines.values():
        spine.set_linewidth(1.7)

plt.tight_layout()
plt.savefig(
    # "Taylor_SkillScore_Regional_v7.png",
    "Taylor_SkillScore_Regional_tak_f2.png",

    dpi=300,
    bbox_inches="tight"
)
plt.show()
```
